Разработать проект для управления растениями в саду, используя основные концепции объектно-ориентированного программирования (ООП) в C#. Вам нужно создать классы, которые будут описывать разные типы растений, используя наследование и полиморфизм.


```
class Plant
{
    private string name;
    private string type;
    private float high;

    public string Name
    {
        get{return name;}
        set{name = value;}
    }

    public string Type
    {
        get{return type;}
        set{type = value;}
    }

    public float High
    {
        get{return high;}
        set{high = value;}
    }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        High = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет!");
    }

    public void Grow(float amount)
    {
        high+=amount;
    }
}

class Flower : Plant
{
    private string color;
    private int count;

    public string Color
    {
        get{return color;}
        set{color= value;}
    }
    public int Count
    {
        get{return count;}
        set {count = value;}
    }
    public Flower(string name, float height, string color, int count) : base(name, "Цветок", height)
    {
        Color = color;
        Count = count;
    }
    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }
    public void Fading()
    {
        Console.WriteLine("Цветок увядает");
    }
}
class Tree : Plant
{
    private int age;
    private string typeLeaves;
    
    public int Age
    {
        get {return age;}
        set{age = value;}
    }

    public string TypeLeaves
    {
        get{return typeLeaves;}
        set{typeLeaves = value;}
    }

    public Tree(string name, float height, int age, string typeLeaves) : base(name, "Дерево", height)
    {
        Age = age;
        TypeLeaves = typeLeaves;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} стало старше на {years} лет. Текущий возраст: {Age} лет");
    }
}
Flower myFlower = new Flower("Хризантема", 20, "Красный", 4);
Tree myTree = new Tree("Сосна", 150, 8, "Хвойные");

Console.WriteLine("===Цветок===");
Console.WriteLine($"Растение: {myFlower.Name}, Тип: {myFlower.Type}, Высота: {myFlower.High} см, Цвет: {myFlower.Color}, Количество лепестков: {myFlower.Count}");
myFlower.Bloom();
myFlower.Grow(25);
myFlower.Fading();

Console.WriteLine("===Дерево===");
Console.WriteLine($"Растение: {myTree.Name}, Тип: {myTree.Type}, Высота: {myTree.High} см, Возраст: {myTree.Age} лет, Тип листьев: {myTree.TypeLeaves}");
myTree.Bloom();
myTree.Grow(30);
myTree.GrowOlder(2);
```
