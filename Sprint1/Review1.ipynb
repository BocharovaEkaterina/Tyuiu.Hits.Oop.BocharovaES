Создайте проект на C#, реализующее классы Товар (Product), Продавец (Seller), Магазин (Store) с использование геттеров и сеттеров, а также модификаторов доступа public, private, protected, internal. От каждого класса создайте по одному экземпляру.  

```
using System;
using System.Collections.Generic;

class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }
    public decimal Price
    {
        get { return price; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Цена не может быть отрицательной!");
            price = value;
        }
    }
    public int Quantity
    {
        get { return quantity; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Количество не может быть отрицательным!");
            quantity = value;
        }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }
    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Количество: {Quantity}, Категория: {Category}, Описание: {Description}";
    }
}


class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Зарплата не может быть отрицательной!");
            salary = value;
        }
    }
    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Товар {product.Name} добавлен продавцу.");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"Продано товара {product.Name}: {quantity} шт.");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара {product.Name} на складе!");
        }
    }

    public string GetSellerInfo()
    {
        return $"Имя: {Name}, Табельный номер: {EmployeeId}, Зарплата: {Salary} руб., Контакт: {ContactInfo}";
    }
}


class Store
{
    private string storeName;
    private string location;
    public string storeHours;

    private List<Seller> sellers = new List<Seller>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }
    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин.");
    }

    public void ListProducts()
    {
        Console.WriteLine("Товары продавцов:");

        foreach (Seller seller in sellers)
        {
            Console.WriteLine($"Товары продавца: {seller.Name}");
        }
    }
    public string GetStoreInfo()
    {
        return $"Название магазина: {StoreName}, Адрес: {Location}, Часы работы: {StoreHours}";
    }
}
Product product = new Product("Ноутбук", 75000, 5, "Электроника", "Ноутбук для работы и учебы");
Seller seller = new Seller("Екатерина Сергеевна","S102",80000,"+7 900 000-55-55");
Store store = new Store("DNS","ул. Тимофея Чаркова, 60","10:00 - 22:00");

Console.WriteLine("=== ТОВАР ===");
Console.WriteLine(product.GetProductInfo());
Console.WriteLine($"Общая стоимость: {product.GetTotalPrice()} руб.");
product.UpdateQuantity(2);
Console.WriteLine($"Количество после изменения: {product.Quantity}");
Console.WriteLine();

Console.WriteLine("=== ПРОДАВЕЦ ===");
seller.AddProduct(product);
Console.WriteLine(seller.GetSellerInfo());
seller.SellProduct(product, 1);
Console.WriteLine($"Количество товара после продажи: {product.Quantity}");
Console.WriteLine();

Console.WriteLine("=== МАГАЗИН ===");
store.AddSeller(seller);
Console.WriteLine(store.GetStoreInfo());
store.ListProducts();
```
