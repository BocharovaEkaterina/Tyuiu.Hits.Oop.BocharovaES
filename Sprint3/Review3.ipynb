Задание
Создайте программу, демонстрирующую принципы полиморфизма, перегрузки и перекрытия методов в контексте класса "Компьютер" и его компонентов. Используйте обобщенные типы (Generic) для создания гибкой системы.

```
using System;
using System.Collections.Generic;

public abstract class Component
{
    public string Name{ get; set;}
    private double _price;

    public double Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не может быть отрицательной!");
        }
    }

    public Component(string name, double price)
    {
        Name=name;
        Price= price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, Цена: {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Name} ({Price} руб.)";
    }
}

public interface IInstallable
{
    void Install();
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, double price, int cores) : base(name, price)
    {
        Cores = cores;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Процессор: {Name}, Ядра: {Cores}, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен в сокет.");
    }
    public static bool operator ==(CPU a, CPU b)
    {
        if (a is null && b is null) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is CPU other)
            return this.Name == other.Name;
        return false;
    }

    public override int GetHashCode()
    {
        if (Name != null)
        {
            return Name.GetHashCode();
        }
        else
        {
            return 0;
        }
    }
}

public class RAM : Component, IInstallable
{
    public int SizeGB { get; set; }

    public RAM(string name, double price, int sizeGb) : base(name, price)
    {
        SizeGB = sizeGb;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Оперативная память: {Name}, Объем: {SizeGB} GB, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Оперативная память {Name} установлена в слот.");
    }
    public static bool operator ==(RAM a, RAM b)
    {
        if (a is null && b is null) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name;
    }
    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is RAM other)
            return this.Name == other.Name;
        return false;
    }

    public override int GetHashCode()
    {
        if (Name != null)
        {
            return Name.GetHashCode();
        }
        else
        {
            return 0;
        }
    }
}

public class GPU : Component, IInstallable
{
    public int VideoMemory { get; set; }

    public GPU(string name, double price, int videoMemory) : base(name, price)
    {
        VideoMemory = videoMemory;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта: {Name}, Память: {VideoMemory} GB, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлена в слот PCI-E.");
    }
    public static bool operator ==(GPU a, GPU b)
    {
        if (a is null && b is null) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is GPU other)
            return this.Name == other.Name;
        return false;
    }

    public override int GetHashCode()
    {
        if (Name != null)
        {
            return Name.GetHashCode();
        }
        else
        {
            return 0;
        }
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityTB { get; set; }

    public HDD(string name, double price, int capacityTb) : base(name, price)
    {
        CapacityTB = capacityTb;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Жесткий диск: {Name}, Объем: {CapacityTB} TB, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Жесткий диск {Name} установлен в отсек.");
    }
    public static bool operator ==(HDD a, HDD b)
    {
        if (a is null && b is null) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name;
    }
    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is HDD other)
            return this.Name == other.Name;
        return false;
    }

    public override int GetHashCode()
    {
        if (Name != null)
        {
            return Name.GetHashCode();
        }
        else
        {
            return 0;
        }
    }
}
public class Computer
{
    private List<Component> _components = new List<Component>();
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
            Console.WriteLine($"Компонент {component.Name} удален из компьютера.");
        else
            Console.WriteLine($"Компонент {component.Name} не найден.");
    }
    public double GetTotalPrice()
    {
        double total = 0;
        foreach (var c in _components)
        {
            total += c.Price;
        }
        return total;
    }
    public void ShowAllComponents()
    {
        Console.WriteLine("--- Компоненты компьютера ---");
        foreach (var c in _components)
        {
            Console.WriteLine(c.ToString());
        }
    }
}

public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        Computer pc = new Computer();
        pc.AddComponent(new CPU("Intel Core i7-14700K", 45000, 20));
        pc.AddComponent(new GPU("NVIDIA RTX 4090", 150000, 24));
        pc.AddComponent(new RAM("Kingston Fury 32GB", 12000, 32));
        pc.AddComponent(new HDD("Seagate 2TB", 8000, 2));
        return pc;
    }

    public static Computer CreateOfficePC()
    {
        Computer pc = new Computer();
        pc.AddComponent(new CPU("Intel Core i3", 15000, 4));
        pc.AddComponent(new RAM("Crucial 8GB", 3000, 8));
        pc.AddComponent(new HDD("WD 1TB", 4000, 1));
        return pc;
    }
    public static Computer CreateCustomPC<T1, T2, T3>(T1 cpu, T2 gpu, T3 storage)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
        where T3 : Component, IInstallable
    {
        Computer pc = new Computer();
        pc.AddComponent(cpu);
        pc.AddComponent(gpu);
        pc.AddComponent(storage);
        return pc;
    }
}
Console.WriteLine("===РУЧНАЯ СБОРКА КОМПЬЮТЕРА ===");
Computer customPC = new Computer();
customPC.AddComponent(new CPU("AMD Ryzen 5", 20000, 6));
customPC.AddComponent(new RAM("HyperX 16GB", 8000, 16));
customPC.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {customPC.GetTotalPrice()} руб.");
Console.WriteLine();

Console.WriteLine("---Компьютерная фабрика---");
Console.WriteLine();
Console.WriteLine("===СБОРКА ИГРОВОГО КОМПЬЮТЕРА===");
Computer gamingPC = ComputerFactory.CreateGamingPC();
gamingPC.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {gamingPC.GetTotalPrice()} руб.");
Console.WriteLine();

Console.WriteLine("===СБОРКА ОФИСНОГО КОМПЬЮТЕРА ===");
Computer officePC = ComputerFactory.CreateOfficePC();
officePC.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {officePC.GetTotalPrice()} руб.");
Console.WriteLine();

Console.WriteLine("=== СБОРКА КАСТОМНОГО КОМПЬЮТЕРА (ЧЕРЕЗ GENERIC-МЕТОД) ===");

Computer myCustomPC = ComputerFactory.CreateCustomPC(
    new CPU("AMD Ryzen 9 7950X", 60000, 16),
    new GPU("NVIDIA RTX 4080", 120000, 16),
    new RAM("Corsair Vengeance 64GB", 25000, 64)
);

myCustomPC.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {myCustomPC.GetTotalPrice()} руб.");
Console.WriteLine();

Console.WriteLine("=== ДОБАВЛЕНИЕ И УДАЛЕНИЕ ===");
HDD oldDisk = new HDD("Seagate 2TB", 8000, 2); 
gamingPC.RemoveComponent(oldDisk);
HDD newDisk = new HDD("Samsung 4TB", 12000, 4);
gamingPC.AddComponent(newDisk);
gamingPC.ShowAllComponents();
Console.WriteLine();

Console.WriteLine("=== ИНФОРМАЦИЯ О КОМПОНЕНТАХ (ПОЛИМОРФИЗМ) ===");
Component[] myComponents = 
{
    new CPU("AMD Ryzen 7", 35000, 8),
    new GPU("NVIDIA RTX 4090", 150000, 24),
    new RAM("Kingston Fury 32GB", 12000, 32),
    new HDD("Seagate 2TB", 8000, 2)
};

foreach (var comp in myComponents)
{
    comp.GetInfo(); 
}
Console.WriteLine();


CPU cpu1 = new CPU("Intel Core i7", 10000, 4);
CPU cpu2 = new CPU("Intel Core i7", 12000, 4);

Console.WriteLine("--- Демонстрация перегрузки операторов == ---");
Console.WriteLine($"Сравниваем: {cpu1.ToString()} и {cpu2.ToString()}");
Console.WriteLine($"Результат (cpu1 == cpu2): {cpu1 == cpu2}");
```
